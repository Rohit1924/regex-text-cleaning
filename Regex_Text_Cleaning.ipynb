# **Regex Text Cleaning**
**This project uses Python Regular Expressions to clean subtitle text
by removing timestamps and unwanted characters.**


```
## Import all necessary modules
# Import the Google Colab files module
from google.colab import files
```

```
## File upload function
# Upload a file from your local computer to Google Colab
file = files.upload()
```

```
# Check the data type of the uploaded file
print(type(file))
```

```
key = list(file.keys())  # Get the uploaded file name(s) as a list
text = str(file[key[0]])  # Convert the content of the first uploaded file into a string
text  # Display the file content
```

```
## Pattern to replace
import re  # Import the Regular Expression module

'\\n\\n00:02.220 --> 00:08.040\\'  # Example timestamp pattern present in the text

Patter3repalce = r'\\n\\n\d{0,3}\d:\d\d.\d\d\d --> \d\d:\d\d.\d\d\d\\'  # Define a regex pattern to match subtitle timestamps

res_time = re.compile(Patter3repalce)  # Compile the regex pattern for searching

res_time.findall(text)  # Find all timestamp patterns that match the regex in the text
```

```
 # Remove all matched timestamp patterns from the text
newtext = res_time.sub('', text)
```

```
s = re.compile('\'')  # Create a regex pattern to match single quotes
new = s.sub('', newtext)  # Remove all single quotes from the text
newtext = new[12:-3]  # Remove the unwanted characters from the beginning and end of the text
newtext  # Display the cleaned text
```
